```python
from dotenv import load_dotenv
from openrouter import OpenRouter
import os
```

```python
load_dotenv(override=True)

open_router_api_key = os.getenv('OPEN_ROUTER_API_KEY')
model = os.getenv('MODEL', 'openai/gpt-oss-20b:free')

if open_router_api_key:
    print(f"Open Router API Key exists and begins {open_router_api_key[:8]}")
else:
    print("OpenAI API Key not set - please head to the troubleshooting guide in the setup folder")

print(f"Using model: {model}")
```

```python
from IPython.display import Markdown, display

question = "What is the meaning of life?"

response = OpenRouter(api_key=open_router_api_key).chat.send(
    model=model,
    messages=[
        {"role": "user", "content": question}
    ],
)

answer = response.choices[0].message.content
display(Markdown(answer))
```

```python
message = f"""
Here is a question:
{question}

And here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""

messages = [{"role": "user", "content": message}]
response = OpenRouter(api_key=open_router_api_key).chat.send(
    model=model,
    messages=messages,
)

display(Markdown(response.choices[0].message.content))
```
